**5 #** Elabore um algoritmo que calcule e visualize a área da base e o volume de um cilindro, sabendo que para isso é necessário conhecer o raio e a altura do cilindro.

```python
r = float(input("Digite o raio a base do cilindro: "))
al = float(input("Digite a altura do cilindo"))

a_base = 3.14 * (r * r)

print("Área da base:", a_base)

vol = a_base * al

print("Volume do cilindro:", vol)
```

**7 #** Elabore um algoritmo que solicite o tamanho em metros quadrados de uma área a ser pintada. Considere que a cobertura da tinta é de 1 litro para cada 3 metros quadrados e que a tinta é vendida em latas de 18 litros, que custam R$ 92,00. Informe ao usuário na saída a quantidades de latas de tinta a serem compradas e o preço total.

```python
l1 = int(input("Digite o tamanho do primeiro lado: "))
l2 = int(input("Digite o tamanho do segundo lado: "))

a = l1 * l2

t_usada = a // 3

t_comprada = 18 // t_usada

val_t = 92.00 * t_comprada

print(f"Você vai usar {t_usada}L de tinta, então comprará {t_comprada} lata(s), gastando R${val_t}")
```

**10 #** Elabore um algoritmo que leia três números e mostre o maior deles.

```python
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
n3 = int(input("Digite o terceiro número: "))

m_num = 0

if n1 > m_num:
    m_num = n1

if n2 > m_num:
    m_num = n2

if n3 > m_num:
    m_num = n3

print("O maior dos números foi:", m_num)
```

**11 #** Elabore um algoritmo que leia três números e mostre o maior e o menor deles.

```python
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
n3 = int(input("Digite o terceiro número: "))

m_num = 0
m_min = 9999999999999

if n1 > m_num:
    m_num = n1

if n2 > m_num:
    m_num = n2

if n3 > m_num:
    m_num = n3

if n1 < m_min:
    m_min = n1

if n2 < m_min:
    m_min = n2

if n3 < m_min:
    m_min = n3

print("O maior dos números foi:", m_num)
print("O menor dos números foi:", m_min)
```
