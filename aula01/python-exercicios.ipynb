**10 #** Elabore um algoritmo que leia três números e mostre o maior deles.

```python
print("Vamos ler três números e mostrar o maior.")

x = 0

maior_num = 0

while x < 3:
    
    try:
        n = input("Digite um número: ").strip().replace(",", ".")
        n = float(n)

        if n > maior_num:
            maior_num = n

        x += 1
        
    except ValueError:
        print("Digite um valor numérico válido")
        continue

print("O maior dos números foi:", maior_num)
```

**11 #** Elabore um algoritmo que leia três números e mostre o maior e o menor deles.

**18 #** Elabore um algoritmo que leia um grupo de valores reais e determine quantos valores são positivos e quantos são negativos.

```python
print("Digite quantos números você quiser, e eu direi quantos são positivos e quantos são negativos. Caso queira parar, digite um valor não numérico.")

#Criação das listas para armazenar os números positivos e negativos
positivos = []
negativos = []

#Loop para coletar os números do usuário
#Esse while sempre será True, então ele só será interrompido quando o usuário decidir encerrar a coleta de números
while True:

    #O bloco try/except é usado para capturar erros de conversão de tipo, caso o usuário digite um valor não numérico
    try:

        #Solicita ao usuário que digite um número, substituindo vírgulas por pontos para permitir a entrada de números decimais
        #Depois disso, o valor é convertido para float
        n = input("Digite um número: ").replace(",", ".")
        n = float(n)

        #Verifica se o número é positivo ou negativo e adiciona à lista correspondente com o método .append()
        if n > 0:
            positivos.append(n)
        else:
            negativos.append(n)

    #Aqui o except se tornou feature, pois ele captura o erro de conversão de tipo e permite que o usuário decida se quer encerrar a coleta de números ou continuar
    except ValueError:
        print("⚠️Valor não numérico detectado.")
        print("Deseja encerrar a coleta? (S/N)")

        #O input do usuário é capturado, e o método .strip() remove espaços em branco no início e no final da string, enquanto o método .upper() converte a string para maiúsculas,
        #facilitando a comparação
        encerrar = input("-> ").strip().upper()

        #Se o usuário digitar "S", o loop é interrompido com a instrução break. Caso contrário, uma mensagem é exibida informando que a coleta de números continuará, 
        #e o loop continua com a instrução continue
        if encerrar == "S":
            break
        else:
            print("Continuando a coleta de números. Digite um número válido.")
            continue

#Após o loop, uma visão geral dos números coletados é exibida, mostrando a quantidade de números positivos e negativos armazenados nas listas correspondentes, com a ajuda
#da função len() que retorna o tamanho da lista
print("========================================")
print("Visão geral dos números coletados:")
print("➕Números positivos:", len(positivos))
print("➖Números negativos:", len(negativos))
```

**19 #** Elabore um algoritmo que leia o sexo e a altura de 10 pessoas e mostre a altura média das mulheres e dos homens separadamente.

```python
print("Agora, vamos coletar a altura de 10 pessoas, e calcular a média de altura entre homens e mulheres, de maneira separada.")

#Como o número de pessoas é fixo, podemos usar um loop while com uma variável de controle x que será incrementada a cada iteração até atingir 10
x = 0

#Criação das listas para armazenar as alturas de homens e mulheres
homens = []
mulheres = []

#Loop para coletar a altura e o sexo das pessoas
while x < 10:

    #Solicita ao usuário que digite o sexo da pessoa, removendo espaços em branco e convertendo para maiúsculas para facilitar a comparação com o uso do método .strip() e .upper()
    sexo = input("Digite o sexo (H/M): ").strip().upper()

    #Tratativa de erro para garantir que o usuário digite apenas "H" ou "M". Caso contrário, uma mensagem de erro é exibida e o loop continua com a instrução continue
    if sexo not in ("H", "M"):
        print("Sexo inválido! Digite H ou M.")
        continue

    #Solicita ao usuário que digite a altura da pessoa, substituindo vírgulas por pontos para permitir a entrada de números decimais
    al = input("Digite a altura (Em metros, ex: 1,85m; 1,90m): ").replace(",", ".")
    
    #Tenta tornar a variável referente à altura em um float e em caso de tratativa de erro, exibe mensagem e o loop continua com a instrução continue
    try:
        al = float(al)
    except ValueError:
        print("⚠️Valor inválido, digite um valor numérico (E em metros)")
        continue

    #Verifica o sexo da pessoa e adiciona a altura à lista correspondente (homens ou mulheres) com o método .append()
    if sexo == "H":
        homens.append(al)
    else:
        mulheres.append(al)
    
    #Incrementa a variável de controle x para avançar para a próxima pessoa
    x += 1

#Calcula a média de altura entre homens e mulheres, utilizando a função sum() para somar os valores das listas e a função len() para obter o número de elementos em cada lista.
#O resultado é arredondado para duas casas decimais com a função round()
media_homens = round(sum(homens) / len(homens), 2)
media_mulheres = round(sum(mulheres) / len(mulheres), 2)

#Exibe a média de altura entre homens e mulheres, juntamente com a quantidade de pessoas em cada grupo, utilizando a função len() para obter o número de elementos em cada lista
print("Média de altura entre os", len(homens), "homens:", media_homens, "metros")
print("Média de altura entre as", len(mulheres), "mulheres:", media_mulheres, "metros")
```
