## O que é Pandas?

> **Pandas** (*Panel Data*, dados em painel) é uma biblioteca *open-source* que fornece alto desempenho e facilidade de uso para estruturas de dados e ferramentas de análise de dados.

**Pontos-chave:** 
* Objeto DataFrame rápido e eficiente com valores default;
* Ferramentas para carregamento de dados em memória a partir de diferentes formatos;
* Tratamento de dados faltantes (missing);
* Fatiamento e reformatação de grandes conjuntos de dados;
* Colunas podem ser deletadas ou inseridas;
* Agrupamento de dados usando agregação e transformações;
* Alto desempenho em união e junção de dados;
* Habilitado para séries temporais.

### Criando DataFrames

```python
# Criando um DataFrame manual
dados = {
    "Jogador": ["Ana", "Lucas", "Maria", "João", "Carlos", "Ana"],
    "Genero": ["F", "M", "F", "M", "M", "F"],
    "Idade": [19, 23, 19, 19, None, 19],
    "UF": ["SP", "SP", "RJ", "ES", "PE", "SP"],
    "Personagem": ["Mago", "Guerreiro", "Arqueira", "Ladino", "Guerreiro", "Mago"],
    "Nível": [12, 20, 15, 18, 20, 12],
    "Pontos": ["2500", "5000", "3200", "4100", "5000", "2500"],
    "Tempo_horas": ["35.5", "50.2", "42.8", "47.3", "50.2", "35.5"],
    "Data_e_Hora_ult_jogo": ["2025-08-20 14:30:00","2025-08-28 20:15:00","2025-08-25 18:45:00","2025-08-27 10:00:00","2025-08-30 23:10:00","2025-08-20 14:30:00"
    ]
}

df = pd.DataFrame(dados)
df
```

### Tratamentos de Dados

```python
print(df.isnull().sum())
```

```python
df.fillna(df["Idade"].mean(), inplace=True)  # Preencher com média
```

```python
# Verificando se tenho linhas duplicadas
tem_duplicados = df.duplicated().any()
print("Existem linhas duplicadas?", tem_duplicados)
```

```python
# Remove linhas duplicadas considerando todas as colunas
df = df.drop_duplicates()
```

```python
df
```

```python
df['Pontos'].sum()
```

```python
df["Pontos"] = df["Pontos"].astype(int)  # Float -> Int
df["Tempo_horas"] = df["Tempo_horas"].astype(float)  # Float -> Int
df["Data_e_Hora_ult_jogo"] = pd.to_datetime(df["Data_e_Hora_ult_jogo"])
```

```python
df.info()
```

### Criação de Colunas

```python
df["Eficiência"] = round(df["Pontos"] / df["Tempo_horas"], 2)
```

```python
df['Faixa_de_Idade'] = df['Idade'].apply(lambda x: 'menos de 20 anos' if x < 20 else 'a partir de 20 anos')
```

```python
# Renomeando a coluna
df.rename(columns={'Faixa_de_Idade': 'Faixa_Idade'}, inplace=True)
```

```python
df.columns
```

```python
df
```

```python
from datetime import datetime

# Data atual
agora = datetime.now()

# Diferença de tempo
df["Dias_sem_jogar"] = (agora - df["Data_e_Hora_ult_jogo"]).dt.days
df["Horas_sem_jogar"] = round((agora - df["Data_e_Hora_ult_jogo"]).dt.total_seconds() / 3600, 2)
```

### Joins (Inner, Left, Right, Full)

**Mesclagem (*merge*)**: conecta linhas em *DataFrames* com base em uma ou mais chaves.

![picture](https://files.realpython.com/media/join_diagram.93e6ef63afbe.png)

```python
df
```

```python
df_itens = pd.DataFrame({
    "Jogador": ["Ana", "Lucas", "Maria", "Pedro"],
    "Item": ["Espada", "Escudo", "Arco", "Cajado"]
})

inner = pd.merge(df, df_itens, on="Jogador", how="inner")
left = pd.merge(df, df_itens, on="Jogador", how="left")
right = pd.merge(df, df_itens, on="Jogador", how="right")
full = pd.merge(df, df_itens, on="Jogador", how="outer")
```
