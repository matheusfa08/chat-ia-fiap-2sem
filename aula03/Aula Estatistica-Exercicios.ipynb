### Exercicio 6 Intreprete os resultado do calculo de correlação Age x Overall e SprintSpeed x Acceleration 

```python
# Correlação entre idade e Overall
corr_idade_overall = df["Age"].corr(df["Overall"])
print("Correlação Age x Overall:", corr_idade_overall)

# SprintSpeed x Acceleration
corr_speed_acc = df["SprintSpeed"].corr(df["Acceleration"])
print("Correlação SprintSpeed x Acceleration:", corr_speed_acc)
```

### Exercicio 7 Intreprete os resultado e responda, Se um joagodor tiver 80 em SprintSpeed quanto tera em Acceleration?

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Scatter plot simples: SprintSpeed x Acceleration
plt.figure(figsize=(8,6))
sns.scatterplot(x="SprintSpeed", y="Acceleration", data=df)
plt.title("Correlação: Sprint Speed x Acceleration")
plt.xlabel("Sprint Speed")
plt.ylabel("Acceleration")
plt.show()

# Scatter plot com linha de tendência (regressão)
plt.figure(figsize=(8,6))
sns.regplot(x="SprintSpeed", y="Acceleration", data=df, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title("Correlação: Sprint Speed x Acceleration (com tendência)")
plt.xlabel("Sprint Speed")
plt.ylabel("Acceleration")
plt.show()

# Cálculo da correlação
corr = df["SprintSpeed"].corr(df["Acceleration"])
print("Correlação entre Sprint Speed e Acceleration:", corr)
```
